### ACCIDENT PREDICTION

where is the most dangerous spot to travel to

- based on most dangerous and least dangerous location -- location risk prediction
- using the area, accident spot columns for averaging the dangerous places rather than counties 

```python
from sklearn.model_selection import train_test_split, GridSearchCV, RandomizedSearchCV
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, roc_curve, auc
```

```python
def get_models():
    models = {
        'Logistic Regression': LogisticRegression(solver='lbfgs', C=10),
        'Decision Tree': DecisionTreeClassifier(criterion='entropy', max_depth=10, min_samples_split=5),
        'Random Forest': RandomForestClassifier(n_estimators=100, max_depth=None),
        'KNN': KNeighborsClassifier(n_neighbors=5, weights="uniform", metric="euclidean")
    }
    return models
```

#### PREDICTING MOST/LEAST DANGEROUS COUNTIES (LOCATION RISK)

- calculated the average victim deaths per accident spot.
- a death casualty threshold so that i can categorize into low and high
- create the feature targets and split the data into training and test datasets
- 

```python
def calculate_location_risk():
    # calculate average deaths per area
    area_avg_deaths = df.groupby('Accident Spot')[
        'Dead Casualties Count'].mean()
    death_thresholds = area_avg_deaths.quantile([0.22, 0.78])

    df['Location Risk'] = df['Accident Spot'].map(
        lambda x: 'Low' if area_avg_deaths[x] <= death_thresholds.iloc[0]
        else 'High'
    )
    return df


calculate_location_risk()
```

```python
def create_feature_target(df):
    features = [
        'Accident Spot',
        'Area',
    ]

    X = df[features]
    y_deaths = pd.cut(
        df['Dead Casualties Count'],
        bins=[1, 3, np.inf],
        labels=['Low', 'High']
    )

    y_location = df['Location Risk']
    return X, y_deaths, y_location


X, y_deaths, y_location = create_feature_target(df)
df.head()
```

```python
def create_preprocessor():
    categorical_features = ['Accident Spot', 'Area']

    categorical_transformer = OneHotEncoder(handle_unknown='ignore')

    preprocessor = ColumnTransformer(
        transformers=[
            ('cat', categorical_transformer, categorical_features)
        ]
    )
    return preprocessor


def split_data(X, y, test_size=0.35, random_state=45):
    return train_test_split(X, y, test_size=test_size, random_state=random_state)

# model training and evaluation


def train_evaluate_models(X, y, preprocessor):
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.35, random_state=45)
    models = get_models()
    results = {}

    for name, model in models.items():
        pipeline = Pipeline([
            ('preprocessor', preprocessor),
            ('classifier', model)
        ])

        pipeline.fit(X_train, y_train)
        y_pred = pipeline.predict(X_test)
        accuracy = accuracy_score(y_test, y_pred)
        results[name] = accuracy
        print(f"{name} - Accuracy: {accuracy:.3f}")
        print(classification_report(y_test, y_pred, zero_division=1.0))
        print("-----------------------------")

    return results


def main():
    preprocessor = create_preprocessor()
    print("\nTraining models for Location Risk Prediction:")
    results_location = train_evaluate_models(X, y_location, preprocessor)


if __name__ == '__main__':
    main()
```
